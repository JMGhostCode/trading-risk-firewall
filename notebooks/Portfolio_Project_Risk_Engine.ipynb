```python
import pandas as pd
import numpy as np
import random

# Configuración básica
num_trades = 120
fechas = pd.date_range(start="2023-09-01", periods=30, freq='B') # Días laborables simulados

datos = []
for i in range(num_trades):
    fecha = random.choice(fechas).strftime("%d/%m/%Y")

    # Generar hora aleatoria entre 09:00 y 15:59
    hora = f"{random.randint(9, 15):02d}:{random.randint(0, 59):02d}"

    # Probabilidades de aparición
    fase = random.choices(['Fase 1', 'Fase 2', 'Fase 3', 'Fase 4', 'FOBO'], weights=[0.3, 0.1, 0.4, 0.15, 0.05])[0]
    trade = random.choices(['Tipo 1', 'Tipo 2', 'Tipo 3', 'CT'], weights=[0.4, 0.3, 0.2, 0.1])[0]
    posicion = random.choice(['Corto', 'Largo'])
    calidad = random.choices(['A+', 'A', 'B', 'C'], weights=[0.4, 0.3, 0.2, 0.1])[0]

    # Lógica sesgada para que el análisis cuantitativo del portafolio tenga sentido
    prob_exito = 0.4
    if fase in ['Fase 1', 'Fase 4']: prob_exito += 0.2
    if fase == 'Fase 3': prob_exito -= 0.2
    if trade == 'Tipo 1' and posicion == 'Corto': prob_exito += 0.15
    if trade == 'Tipo 2' and posicion == 'Largo': prob_exito += 0.15

    funciono = 'Si' if random.random() < prob_exito else 'No'

    # Generación de valores monetarios y de riesgo
    comision = 1.90
    if funciono == 'Si':
        pnl = round(random.uniform(15.0, 45.0), 2)
        mae = random.randint(1, 7) # Ganadores retroceden poco
        mfe = random.randint(15, 30)
        comentario = random.choice(["Buen trade, se respetaron las reglas.", "Gestión impecable.", "Se dejó correr hasta el target."])
    else:
        pnl = round(random.uniform(-15.0, -25.0), 2)
        mae = random.randint(10, 15)
        mfe = random.randint(1, 5)
        comentario = random.choice(["Entrada tardía por FOMO.", "Reversión inesperada en Fase 3.", "Stop ajustado por ansiedad.", "Rango no detectado a tiempo."])

    datos.append([fecha, hora, fase, trade, posicion, calidad, funciono, pnl, comision, mae, mfe, comentario])

# Creación del DataFrame
columnas = ['Fecha', 'Hora', 'Fases', 'Trade', 'Posición', 'Quality_Setup', 'Funcionó', 'PnL Neto ($)', 'Comision', 'MAE_TICKS', 'MFE_TICKS', 'Comentario']
df_dummy = pd.DataFrame(datos, columns=columnas)

# Exportar archivo
nombre_archivo = 'dummy_data.csv'
df_dummy.to_csv(nombre_archivo, index=False)
print(f"Archivo '{nombre_archivo}' generado con éxito con {num_trades} operaciones simuladas.")
```
